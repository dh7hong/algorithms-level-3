#### 1. enumerate()

```python
# Problem without enumerate()

fruits = ['apple', 'banana', 'cherry']

# traditional way
print("Traditional way without enumerate():")
for i in range(len(fruits)):
    print(f"Index: {i}, Fruit: {fruits[i]}")

# using enumerate()
print("Using enumerate():")
for i, fruit in enumerate(fruits):
    print(f"Index: {i}, Fruit: {fruit}")
```

#### 2. zip()

```python
# zip combines multiple iterables into a single iterable of tuples

names = ["John", "Jane", "Bob"]
scores = [90, 85, 100]

for name, score in zip(names, scores):
    print(name, score)
```

```python
names = ["John", "Jane"]
scores = [90, 85]
ages = [20, 22]

for name, score, age in zip(names, scores, ages):
    print(name, score, age)
```

```python
# what does zip actually create?

names = ["John", "Jane", "Bob"]
scores = [90, 85, 100]

list(zip(names, scores))
```

#### 3. enumerate() + zip()

```python
names = ["John", "Jane", "Bob"]
scores = [90, 85, 100]

for index, (name, score) in enumerate(zip(names, scores)):
    print(index, name, score)
```
